```python
import feedparser
import re
import time
import urllib.parse  # Add this import

def get_arxiv_ids(search_query='"CMS Open Data"', max_results=500, batch_size=100):
    base_url = "https://export.arxiv.org/api/query"
    arxiv_ids = []
    start = 0

    while start < max_results:
        n = min(batch_size, max_results - start)
        url = (
            f"{base_url}"
            f"?search_query={urllib.parse.quote(search_query)}"  # <-- encode it here
            f"&searchtype=all"
            f"&start={start}"
            f"&max_results={n}"
        )

        feed = feedparser.parse(url)

        if not feed.entries:
            break

        for entry in feed.entries:
            raw_id = entry.id.split("/abs/")[-1]
            arxiv_id = re.sub(r"v\d+$", "", raw_id)
            arxiv_ids.append(arxiv_id)

        start += len(feed.entries)

        if start < max_results:
            time.sleep(3)

    return arxiv_ids


if __name__ == "__main__":
    ids = get_arxiv_ids()
    print(f"Found {len(ids)} arXiv IDs:\n")
    print(ids)
```
